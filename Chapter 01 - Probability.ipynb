**Exercise 1.18.** Suppose that $A_1, \dots A_k$ are a partition of the sample space and that $\P(B) > 0$.

If $\P(A_1 | B) < \P(A_1)$, suppose that $\P(A_i | B) \le \P(A_i)$ for all $i \ne 1$.

Then $1 = \sum_{i=1}^k \P(A_i|B) < \sum_{i=1}^k \P(A_i) = 1$.

Hence $P(A_i | B) > \P(A_i)$ for some $i\ne 1$.

This means that, if $A_1$ has strictly smaller probability given $B$, some other $A_i$ has strictly larger probability given $B$.

**Exercise 1.20.** We are given 5 coins with probability of heads appearing when flipped $p_1,\dots,p_5$ as given in the code in the next cell.

(a) If a random coin is chosen and tossed, then heads is shown, find the posterior probability coin $i$ was chosen

1 - 0

2 - 1/10

3 - 1/5

4 - 3/10

5 - 2/5

(b) If we toss the coin again, what is the probability of seeing another head?

$\P(H_2|H_1) = \sum_{i=1}^5 \P((H_2|C_i)|H_1) \P(C_i|H_1) =  \sum_{i=1}^5 \P(H_2|C_i) \P(C_i|H_1)$

3/4

(c) Find $\P(C_i|B_4)$. The event $B_4$ means that the first head is on toss 4

1 - 0

2 - 27/46

3 - 8/23

4 - 3/46

5 - 0

```python
p = [0,0,f(1,4),f(1,2),f(3,4),f(1,1)]
#(a) P(C_i|H) = P(H|C_i)P(C_i)/P(H)
denom = sum(f(1,5) * p[i] for i in range(1,6))
for i in range(1,6):
    num = f(1,5) * p[i]
    print('(a)', i, ':', num/denom)
```

```python
c = [0,0,f(1,10),f(1,5),f(3,10),f(2,5)]
total = 0
for i in range(6):
    total += c[i] * p[i]
print('(b)', total)
```

```python
# (c) P(C_i | B_4) = P(B_4 | C_i) * P(C_i) / P(B_4)

#P(B_4)
denom = f(1,5) * sum( (1-prob)**3 * prob for prob in p)

for i in range(1,6):
    num = f(1,5) * (1-p[i])**3 * p[i]
    print(f'(c) {i}:', num/denom)
```

**Exercise 1.21.** Proportion of heads as a function of $n$

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)

x = rng.binomial(1,.3,1000)
y = np.arange(1,1001)
z = x.cumsum() / y

plt.plot(y,z)
plt.plot(y, [.3]*1000)
plt.show()
```

```python
x = rng.binomial(1,.03,1000)
y = np.arange(1,1001)
z = x.cumsum() / y

plt.plot(y,z)
plt.plot(y, [.03]*1000)
plt.show()
```
