*   Our first inference problem is nonparametric estimation of the CDF.
*   We then estimate statistical functionals (functions of the CDF), e.g., mean, variance, correlation. This nonparametric method for estimating functionals is called the plug-in method.

# 7.1 $\:$ The Empirical Distribution Function

Let $X_1,\dots,X_n \sim F$ be IID. The *empirical distribution function* $\hat F_n$ is an estimator for $F$. We simply define $\hat F_n(x)$ to be the fraction of observed values $\le x$:
$$ \hat F_n(x) = \frac 1n \sum_{i=1}^n \1(X_i \le x). $$

Most of the following follow from $\1(X_i \le x) \sim \Ber(F(x))$.

**Expectation.** $\hat F_n(x)$ is an unbiased estimator of $F(x)$.

**Variance.** $\var \hat F_n(x) = F(x)(1-F(x))/n$.

**Mean squared error.** $\mse$ is equal to the variance.

**Consistency.** $\hat F_n(x)$ is consistent.

**Glivenko–Cantelli theorem.** Rather than pointwise convergence, consider $\sup_x |\hat F_n(x) - F(x) |$. The Glivenko–Cantelli theorem says that this converges to zero almost surely.

**Dvoretzky–Kiefer–Wolfowitz inequality.** We know that $\sup_x |\hat F_n(x) - F(x) |$ converges to zero in probability. The Dvoretzky–Kiefer–Wolfowitz inequality quantifies this:
$$ \P( \sup_x |\hat F_n(x) - F(x) | > \varepsilon) \le 2 e^{-2n\varepsilon^2}. $$
Note that without $\sup_x$ this would just be Hoeffding's inequality.

**Confidence band for the CDF using the Dvoretzky–Kiefer–Wolfowitz inequality.** This is an exact confidence band. Define
$$
\begin{align*}
L(x) &= \max(\hat F_n(x) - \varepsilon_n, 0),\\
U(x) &= \min(\hat F_n(x) + \varepsilon_n, 1),\\
\varepsilon_n &= \sqrt{\frac {1}{2n} \log \left ( \frac 2 \alpha\right) }.
\end{align*}
$$
Then
$$ \P( F(x) \in [L(x), U(x)] \text{ for all } x \in \R) \ge 1 - \alpha. $$

# 7.2 $\:$ Statistical Functionals

A statistical functional $T(F)$ is a function of a CDF $F$.

**Plug-in estimator.** For $\theta = T(F)$, the plug-in estimator is $\hat \theta_n = T(\hat F_n)$.

**Linear functional.** A functional of the form $T(F) = \int r(x) dF(x)$.

**Plug-in estimator for a linear functional.**

$$ T(\hat F_n) = \int r(x) d\hat F_n(x) = \frac 1n \sum_{i=1}^n r(X_i).$$

Assume we can find an estimated standard error $\se$ of $T(\hat F_n)$. In many cases, $T(\hat F_n) \approx N(T(F), \hat \se^2)$. The approximate normal-based $1-\alpha$ confidence interval for $T(F)$ is then $T(\hat F_n) \pm z_{\alpha/2}\hat \se$.

**Exercise 7.3.** Generate 100 observations from an $N(0,1)$ distribution.

```python
import numpy as np
from scipy.stats import norm, cauchy

rng = np.random.default_rng(0)
X = rng.normal(loc=0,scale=1,size=100)

epsilon = np.sqrt(np.log(2/.05) / (2 * 100))
```

The 100 $N(0,1)$ observations are stored in `X`. With probabililty $\ge 0.95$, for all $x$ we have that $\hat F_n(x)$ is within `epsilon` of $\Phi(x).$ To verify that this is the case, it suffices to check $\hat F_n(x)$ is within `epsilon` of $\Phi(x)$ when $x$ is an observation in `X` or just before (this idea is from GPT).

```python
Xs = np.sort(X)

within_band = True

for i in range(100):
    true = norm.cdf(Xs[i])
    if np.abs(true - i/100) > epsilon or np.abs(true - (i+1)/100) > epsilon:
        within_band = False

if within_band:
    print('The confidence band contains Phi.')
else:
    print('The confidence band does not contain Phi.')
```

Repeat this 1000 times.

```python
hits = 0

for _ in range(1000):
    X = rng.normal(size = 100)
    Xs = np.sort(X)

    within_band = True

    for i in range(100):
        true = norm.cdf(Xs[i])
        if np.abs(true - i/100) > epsilon or np.abs(true - (i+1)/100) > epsilon:
            within_band = False
    if within_band:
        hits+=1

print('The proportion of times that the confidence band contains Phi is', hits/1000)
```

Repeat this using the Cauchy distribution.

```python
hits = 0

for _ in range(1000):
    X = rng.standard_cauchy(size = 100)
    Xs = np.sort(X)

    within_band = True

    for i in range(100):
        true = cauchy.cdf(Xs[i])
        if np.abs(true - i/100) > epsilon or np.abs(true - (i+1)/100) > epsilon:
            within_band = False
    if within_band:
        hits+=1

print('The proportion of times that the confidence band contains cauchy_cdf is', hits/1000)
```

As an aside, note that confidence intervals obtained through Hoeffding's inequality tend to be conservative. The DKW inequality, though formally similar to Hoeffding's inequality, has produced confidence bands above that have coverage close to 0.95.  In general, confidence bands obtained through the DKW inequality will have coverage close to $1-\alpha$.

In the case of redoing the experiment 1000 times with 100 $N(0,1)$'s, the coverage is 0.942. If we view the confidence band of each experiment containing the true CDF roughly as $\Ber(0.95)$, the coverage has standard devation $\sqrt{\frac{0.95 \cdot 0.05}{1000}}\approx 0.006892$. Then $0.942$ is 1.16 standard deviations from $0.95$.

**Exercise 7.4.** Let $X_1,\dots, X_n \sim F$ be IID with empirical distribution function $\hat F_n(x)$. Then

This exercise concerns the pointwise estimator $\hat F_n(x)$. Note that we may view the empirical distribution function as a stochastic process; Donsker’s theorem gives a functional central limit theorem for the empirical CDF.

**Exercise 7.5.** Suppose $x < y$. We compute $\cov(\hat F_n(x), \hat F_n(y)).$ We know that $\E \hat F_n(x) = F(x) $ and $\E \hat F_n(y) = F(y)$. Now,

$$
\begin{align*}
\E (\hat F_n(x) \hat F_n(y) ) &=
\E \left ( \frac{1}{n^2}\sum_{i,j=1}^n \1(X_i \le x)  \1(X_j \le y)\right ) \\
&= \frac{1}{n^2}\E \left ( \sum_{i=1}^n \1(X_i \le x)   + \sum_{i\ne j} \1(X_i \le x)  \1(X_j \le y) \right )\\
&= \frac 1n F(x) + \frac{1}{n^2} \cdot n (n-1) F(x) F(y) \\
&= \frac 1n F(x) + \frac{n-1} n F(x) F(y).
\end{align*}
$$
Then
$$
\begin{align*}
\cov(\hat F_n(x), \hat F_n(y) ) &=
\frac 1n F(x) + \frac{n-1} n F(x) F(y) - F(x) F(y) \\
&= \frac{F(x)(1-F(y))}{n}.
\end{align*}
$$


**Exercise 7.6.** Let $X_1,\dots,X_n \sim F$ be IID with empirical distribution function $\hat F$. Fix $a < b$. Let $\theta = T(F) = F(b) - F(a)$. Then $\hat \theta = T(\hat F) = \hat F(b) - \hat F(a).$ We have
$$
\var(\hat \theta ) = \var \hat F(b) + \var \hat F(a) - 2 \cov(\hat F(b), \hat F(a)) = \frac{F(b)(1-F(b))}{n} + \frac{F(a)(1-F(a))}{n} - \frac{2 F(a) (1-F(b))}{n},
$$
so
$$
\hat \se = \sqrt{\frac{(\hat F(b)-\hat F(a))(1-(\hat F(b)-\hat F(a)))}{n}} = \sqrt{\frac{\hat \theta(1-\hat\theta)}{n}}.
$$
An approximate $1-\alpha$ confidence interval for $\theta$ is $\hat \theta \pm z_{\alpha/2}\hat\se $.

An observation due to GPT is that
$\hat \theta = \frac{1}{n}\sum_{i=1}^n \1(a < X_i\le b)$, which is an average of IID Bernoulli.

**Exercise 7.7.** Data on the magnitudes of earthquakes near Fiji are available on the
website for this book.

```python
import pandas as pd

earthquake_url = "https://www.stat.cmu.edu/~larry/all-of-statistics/=data/fijiquakes.dat"
earthquake = pd.read_csv(earthquake_url, sep=r"\s+")
```

```python
pd.DataFrame(earthquake.mag.describe()).T
```

```python
X = earthquake.mag.to_numpy()
Xs = np.sort(X)
```

```python
n = X.shape[0]
epsilon = np.sqrt(np.log(2/.05) / (2 * n))

def F(x):
    return np.searchsorted(Xs, x, side='right') / n

def U(x):
    return np.minimum(F(x) + epsilon, 1)

def L(x):
    return np.maximum(F(x)-epsilon, 0)
```

```python
import matplotlib.pyplot as plt

dom = np.linspace(3.5,7.0,10_000)
plt.plot(dom, U(dom), label=r'$U(x)$')
plt.plot(dom, F(dom), label=r'$\hat{F}(x)$')
plt.plot(dom, L(dom), label=r'$L(x)$')
plt.legend()
plt.show()
```

To compute an approximate 0.95 confidence interval for $F(4.9) - F(4.3)$, we use the observation in Exercise 7.6. Let $\theta = F(4.9) - F(4.3).$

```python
hat_theta = F(4.9) - F(4.3)
hat_se = np.sqrt(hat_theta*(1-hat_theta)/n)
print('An approximate 0.95 confidence interval for theta is', hat_theta, r'\pm', round(norm.ppf(0.975) * hat_se,3))
```

**Exercise 7.9.** 100 people are given a standard antibiotic to treat an infection and
another 100 are given a new antibiotic.
*   Let $p_1$ be the probability
of recovery under the standard treatment.
*   Let $p_2$ be the probability of
recovery under the new treatment.
*   Let $\theta = p_1-p_2$.

In the first group, 90 people recover, and in the second group, 85 people recover.

```python
hat_p1 = 0.90
hat_p2 = 0.85
hat_theta = 0.05
```

$$\var\hat\theta = \var\hat p_1 + \var \hat p_2 $$

```python
var_hat_theta = hat_p1*(1-hat_p1)/100 + hat_p2*(1-hat_p2)/100
hat_se = np.sqrt(var_hat_theta)
```

```python
print('A 0.80 confidence interval for theta is',hat_theta,r'\pm',round(hat_se * norm.ppf(0.90),4))
print('A 0.95 confidence interval for theta is',hat_theta,r'\pm',round(hat_se * norm.ppf(0.975),4))
```
