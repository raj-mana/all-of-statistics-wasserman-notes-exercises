# 5.1 $\:$ Introduction
**Main ideas.** Given
*   IID sequence $X_1, X_2,\dots$
*   Sample average $\bar X_n$
*   $\mu = \E X_1$
*   $\sigma^2 = \var X_1$

We have:
*   Law of large numbers: $\bar X_n$ converges in probability to $\mu$
*   Central limit theorem: $\frac{\sqrt n}\sigma(\bar X_n - \mu)$ converges in distribution to $N(0,1)$, that is, the sample average is approximately normal if $n$ is large



# 5.2 $\:$ Types of Convergence

**Convergence in distribution.** $X_n \rightsquigarrow X$ if $F_n(t) \to F(t)$ for any point $t$ at which $F$ is continuous. Here $F_n$ and $F$ are   the CDFs of $X_n$ and $X$.
1.   $\liminf \P(X_n \in U) \ge \P(X \in U)$ for any open set $U$
2.   $\limsup \P(X_n \in A) \le \P(X \in A)$ for any closed set $A$
3.   $\lim \P(X_n \in B) = \P(X \in B)$ if $B$ is a continuity set of $B$. This means that the probability $X$ is in the topological boundary of $B$ is zero

To remember the first two bullets, think of point masses that converge in distribution to a point mass. For (1), let $U = (-1,1)$ and consider a sequence of point masses in $U$ that coverge to a point mass at $1$. For (2), let $A$ be a the point mass at $0$ and consider a sequence of point masses that converge to $A$.

**Relationships**
*   Convergence in $L^2$ implies convergence in probability
*   Convergence in probability implies convergence in distribution
*   Convergence in distribution to a constant implies convergence in probability
*   $L^p$ convergence, for any $p > 1$, implies $L^1$ convergence
*   $L^1$ convergence implies convergence in probability
*   Almost-sure convergence implies convergence in probability

**Sums, products, and continuous functions**
*   Sums work as expected for convergence in $L^2$ and probability
*   Products for convergence in probability
*   Compositions for convergence in probability and distribution

**Slutzky's theorem**
*   For the special case that $X_n \rightsquigarrow c$ (a constant), if $Y_n \rightsquigarrow Y$ then
    *   $X_n + Y_n \rightsquigarrow c + Y$
    *   $X_nY_n \rightsquigarrow cY$





# 5.4 $\:$ The Central Limit Theorem
**Sample variance.** Let $X_1, X_2, \dots$ be IID. If we use the sample variance
$$ S_n^2 = \frac{1}{n-1} \sum_{i = 1}^n (X_i - \bar X_n)^2, $$
the central limit theorem is still true:
$$ \frac{\sqrt n}{S_n}(\bar X_n - \mu) \rightsquigarrow N(0,1).$$

**Berry–Essèen theorem.** This says that, for an $L^3$ random variable $X_1$, the CDF of the normal approximation $Z_n = \frac{\sqrt n}{\sigma}(\bar X_n - \mu)$ converges to $\Phi$ in $L^\infty$ with $n^{-1/2}$ decay:
$$\sup_z | \P(Z_n \le z) - \Phi(z) | \le \frac{33}{4} \frac{\E | X_1 - \mu |^3}{\sigma^3} n^{-1/2}.$$
The quantity
$$ \frac{\E |X_1 - \mu|^3}{\sigma^3} $$
is called the standardized ($\sigma$ in denominator) third absolute central ($X_1 - \mu$ has mean zero) moment.

**Multivariate central limit theorem.** Let $X_1, X_2, \dots$ be a sequence of IID random vectors, that is, $X_i = (X_{1i}, \dots, X_{pi})^T$. Let $\mu = \E X_1$ and $$\Sigma = \var X_1 = \E (X-\mu)(X-\mu)^T = \E XX^T - \mu\mu^T.$$ Then
$$ \sqrt n (\bar X_n - \mu) \rightsquigarrow N(0,\Sigma). $$

# 5.5 $\:$ The Delta Method

Suppose that $Y_n$ has limiting normal distribution, that is,
$$ \sqrt n (Y_n - \mu) \rightsquigarrow N(0, \sigma^2). $$
If $g$ is differentiable, then $g(Y_n)$ is approximately normal with
$$ \sqrt n (g(Y_n) - g(\mu) ) \rightsquigarrow N(0, \sigma^2| g'(\mu)|^2).$$
Heuristically, $\sqrt n Y_n$ has standard deviation $\approx \sigma$ and $\sqrt n g(Y_n)$ has standard deviation $\approx \sigma |g'(\mu)| $.

**Multivariate delta method.** If $Y_n$ is a random vector with limiting normal distribution, that is,
$$ \sqrt n (Y_n - \mu) \rightsquigarrow N(0, \Sigma)$$
and $g : \R^p \to \R$ is differentiable, then
$$ \sqrt n(g(Y_n) - g(\mu)) \rightsquigarrow N(0, (\nabla g(\mu))^T\Sigma \nabla g(\mu)).$$

**Exercise 5.1.** Let $X_1, \dots, X_n$ be IID with sample variance $S_n^2$.

(a) The expectation of the sample variance is equal to $\sigma^2$.

$$ \begin{align*}
S_n^2 &= \frac{1}{n-1} \sum_{i=1}^{n} (X_i - \bar X_n)^2 \\
&= \frac{1}{n-1} \sum_{i=1}^{n} (X_i^2  + \bar X_n^2 - 2 X_i \bar X_n) \\
&= \frac{1}{n-1} (I + II + III) \\
\\
\E\left( \frac{1}{n-1} I \right) &= \frac{n}{n-1} (\sigma^2 + \mu^2)\\
\\
II &= \frac 1{n^2} \cdot n \cdot \left( \sum_{j=1}^n X_j \right)^2 \\
&= \frac 1n \left( \sum_{j=1}^n X_j^2 + \sum_{j\ne k} X_j X_k\right) \\
&= (\sigma^2 + \mu^2) + \frac 1n \sum_{i\ne j} X_i X_j \\
\E \left (\frac 1{n-1} II \right) &= \frac{1}{n-1} (\sigma^2 + \mu^2) + \frac{n(n-1)}{n(n-1)} \mu^2 \\
&= \frac{1}{n-1} (\sigma^2 + \mu^2) + \mu^2 \\
\\
III &= - \frac 2n \sum_{i=1}^n \sum_{j=1}^n X_i X_j \\
&= -\frac 2n \sum_{i=1}^n X_i^2 - \frac{2}{n} \sum_{i\ne j} X_i X_j \\
\E\left ( \frac{1}{n-1} III\right ) &= - \frac 2{n-1} (\sigma^2 + \mu^2) - \frac{2n(n-1)}{n(n-1)} \mu^2 \\
&= - \frac 2{n-1} (\sigma^2 + \mu^2) - 2 \mu^2 \\
\\
\E(S_n^2) &= \frac{n}{n-1} (\sigma^2 + \mu^2) + \frac{1}{n-1} (\sigma^2 + \mu^2) + \mu^2 - \frac 2{n-1} (\sigma^2 + \mu^2) - 2\mu^2 \\
&= \sigma^2
\end{align*}$$

(b) The sample variance converges to $\sigma^2$ almost surely. We can use the law of large numbers.
*   $\frac{1}{n-1} I$ converges to $\sigma^2 + \mu^2$ almost surely.
*   $\frac{1}{n-1} II $ converges to $\mu^2$ almost surely.
*   We write $\frac{1}{n-1} III $ as
$$ - 2 \left( \frac{1}{n-1} \sum_{i=1}^n X_i\right ) \bar X_n$$
which converges to $-2 \mu^2$ almost surely.

**Exercise 5.3.** Let $X_1, \dots, X_n$ be IID with finite variance. Then $\bar X_n \to \mu$ in $L^2$.

It suffices to assume that $\mu = 0$. Then
$$\begin{align*}
\bar X_n^2 &= \frac{1}{n^2}\left( \sum_{i=1}^n X_i \right)^2 \\
&=  \frac{1}{n^2}\left( \sum_{i=1}^n X_i^2 + \sum_{i\ne j} X_i X_j \right).
\end{align*} $$
Noting that $\E X_i^2 = \sigma^2$ and $\E \sum_{i\ne j} X_i X_j = \sum_{i\ne j} (\E X_i) (\E X_j) = 0$, it follows that $\E \bar X_n^2 \to 0$.

**Exercise 5.10.** Let $Z \sim N(0,1)$. An application of Markov's inequality gives

**Exercise 5.11.** Let $X_n \sim N(0,1/n)$ and $Z \sim N(0,1)$. Then

**Exercise 5.12.** Let $X, X_1, X_2, \dots$ be positive integer-valued random variables.  

We show that $X_n \rightsquigarrow X$ if and only if
$$ \P(X_n = k) \to \P(X = k)$$
for every integer $k$.

Indeed, if $X_n \rightsquigarrow X$, for each integer $k$ we have that $k + 1/2$ is a continuity point of $F_X$, and so $$F_{X_n}(k) = F_{X_n}(k+1/2) \to F_X(k+1/2) = F_X(k)$$
as $n \to \infty$. Then
$$\P(X_n = k) =  F_{X_n}(k) - F_{X_n}(k-1) \to
F_{X}(k) - F_{X}(k-1) = \P(X =k)$$
as $n \to \infty$.

Conversely, suppose that $\P(X_n = k) \to \P(X =k)$ for each  integer $k$. If $t\in \R$, then
$$
\begin{align*}
F_{X_n}(t) &= \sum_{k = 1}^{\lfloor t \rfloor} \P(X_n = k) \\
&\to \sum_{k = 1}^{\lfloor t \rfloor} \P(X = k) \\
&= F_X(t).
\end{align*}
$$
So $X_n \rightsquigarrow X$.


**Exercise 5.13.** Let $Z_1, Z_2, \dots$ be IID with density $f$ continuous on $[0,\infty)$ and $\P(Z_1 > 0) = 1$. Let $\lambda = \lim_{x \searrow 0} f(x) > 0$. Define
$$ X_n = n \min (Z_1,\dots,Z_n). $$
We show that $X_n \rightsquigarrow Z$ where $Z \sim \Exp(1/\lambda)$.

Indeed, it suffices to show that, for $t > 0$, we have $F_{X_n}(t) \to F_Z(t)$. Let $F(t)= F_{Z_1}(t)$. Now,
$$F_{X_n}(t) = 1 - (1-  F(t/n))^n $$
and $$F_Z(t) = 1- e^{-\lambda t}.$$
It remains to show that
$$ (1 - F(t/n))^n \to e^{-\lambda t}.$$
Equivalently,
$$ n \log (1- F(t/n)) \to -\lambda t.$$
Fixing $t > 0$, using the Taylor expansion of $F$ at $0$ we see that
$$ F(t/n) = \frac{\lambda t}{n} + o(1/n)$$
as $n \to \infty$. Then using the Taylor expansion  $\log(1+x) = x + O(x^2)$ as $x\to 1$, we have
$$ n \log(1 - F(t/n)) = -\lambda t + o(1) \to -\lambda t$$
as $n \to \infty$.
