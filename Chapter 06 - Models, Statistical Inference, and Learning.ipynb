# 6.1 $\:$ Introduction
*Statistical inference*, or *learning*, is the process of using data to *infer the distribution* $F$ that generated the data. We may only want to infer some *feature* of $F$ such as the mean.

# 6.2 $\:$ Parametric and Nonparametric Models
**Statistical model.** A set, denoted $\F$, of *distributions*, *densities*, or *regression functions*.

**Parametric model.** A statistical model that can be parameterized by a *finite number of parameters*. For example, we may write $\F$ as a set of densities $f$:

$$ \F = \{ f(x;\theta) : \theta \in \Theta\}. $$

Here $\theta$ is a single parameter or a vector of parameters that lies in a parameter space $\Theta$. If $\theta$ is a vector but we are only interested in some components of $\theta$, the remaining parameters are called *nuisance parameters*.

**Nonparametric model.** A statistical model that *cannot be parameterized by a finite number of parameters*.

**Example, all CDFs.** The set of all CDFs.

**Example, nonparametric density estimation.** We cannot typically assume all PDFs to form a nonparametric model, but we can choose those PDFs that are smooth with square-integrable second derivative.

**Statistical functionals.** A function of a distribution $F$ is called a statistical functional. We may want to *estimate* a statistical functional given data, e.g., $\int x dF(x)$, $\int x^2 dF(x) - (\int x dF(x))^2$, $F^{-1}(1/2)$.

**Regression, prediction, classification.** Suppose that our observations are pairs of data $(X_1, Y_1), \dots, (X_n,Y_n)$.

*   Here $X$ is called a *predictor*, *regressor*, *feature*, or *independent variable*.
*   $Y$ is called the *outcome*, *response*, or *dependent variable*.
*   $r(x) = \E(Y|X=x)$ is called the *regression function*. A collection of regression functions is called a *regression model*, which can be parametric or nonparametric.

The goal for predicting $Y$ based on a new $X$ value is called *prediction*. The terminology *classification* is used if $Y$ is discrete. If our goal is to estimate $r$, we call this *regression* or *curve estimation*. We may write regression models as
$$ Y = r(X) + \varepsilon$$
where $\E(\varepsilon|X) = 0$.

**Notation.** If $\F = \{ f(x;\theta) : \theta \in \Theta\}$ is a parametric model, denote probability, expectation, and variance corresponding to the density $f(x;\theta)$ by $\P_\theta$, $\E_\theta$, and $\var_\theta$ (*there is no averaging over $\theta$*).

# 6.3 $\:$ Fundamental Concepts in Inference

**Point estimation.** Single best guess of a quantity, e.g., parameter in a parametric model, CDF, PDF, regression function, or prediction of $Y$ given $X$.

**Notation.** Point estimate of $\theta$ is denoted $\hat \theta$ or $\hat \theta_n$. Remember that $\theta$ is fixed and $\hat \theta$ is a random variable.

Formally, if $X_1,\dots,X_n$ are IID, a *point estimator* $\hat \theta $ of a parameter $\theta$ is a function of $X_1,\dots, X_n$.

The bias of an estimator is defined by

$$ \bias(\hat \theta_n) = \E_\theta(\hat\theta_n) - \theta. $$

We say that $\hat \theta_n$ is *unbiased* if $\bias(\hat \theta_n) = 0 $.

We say that $\hat \theta_n$ is *consistent* if $\hat \theta_n \to \theta$ in probability as $n\to \infty$.

The distribution of $\hat \theta_n$ is given a special name, called the *sampling distribution*. Statistical inference concerns the sampling distributions of estimators and test statistics, rather than just the underlying population distribution (distribution that generated the data).

The standard deviation of $\hat \theta_n$ is called the standard error, denoted by $\se$.

$$\se^2 = \se(\hat \theta_n)^2 = \var_\theta(\hat\theta_n).$$

We usually use an estimator to approximate $\se$, denoted by $\hat \se$.

**Mean squared error.** Denoted by $\mse$. This assesses the average squared devation of a point estimate from its true value:
$$ \mse = \E_\theta (\hat\theta_n - \theta)^2. $$
We can decompose
$$ \mse = \bias(\hat\theta_n)^2 + \var_\theta(\hat\theta_n).$$
This is shown by writing
$$ \E_\theta (\hat\theta_n - \theta)^2 =
 \E_\theta (\hat\theta_n - \E_\theta \hat\theta_n + \E_\theta \hat\theta_n - \theta)^2$$
 then expanding.

 The decomposition of $\mse$ can be used to show that, if $\bias \to 0$ ans $\se \to 0$ as $n\to \infty$, then $\hat\theta_n$ is consistent.

 **Asymptotically normal.** An estimator $\hat\theta_n$ for $\theta$ such that
 $$ \frac{\hat\theta_n - \theta}{\se} \rightsquigarrow N(0,1).$$
 Informally, we write $\hat\theta_n\approx N(\theta, \hat\se^2)$.

**Confidence intervals.** A $1-\alpha$ confidence interval for a fixed parameter theta is a "random" interval $C_n = (a,b)$, where $a$ and $b$ are functions of $X_1,\dots,X_n$, such that $\P_\theta(\theta \in C_n) \ge 1- \alpha $ for all $\theta \in \Theta$.

A confidence interval is not a probability statement about $\theta$.

**Approximate normal-based confidence intervals.** If $\hat\theta_n\approx N(\theta, \hat\se^2)$, let
$$ a_{\pm} = \hat\theta_n \pm z_{\alpha/2} \hat \se,$$
where $\P( |N(0,1)| > z_{\alpha/2}) = \alpha $.
Then
$$ \P(\theta \in (a_-, a_+)) \to 1-\alpha $$
as $n\to \infty$.

**Hypothesis testing.** We start with a default theory, called a *null hypothesis*, denoted $H_0$, and ask if data provides sufficient evidence to reject the null hypothesis. Otherwise, we *fail to reject the null hypothesis*. The *alternative hypothesis* is denoted $H_1$.

For example, if $X_1, \dots, X_n$ are IID $\Ber(p)$, we want to test  $H_0 = (p = 1/2)$. It seems reasonable to reject $H_0$ if $ T = |\hat p - \frac 12|$ is large.

# 6.5 $\:$ Appendix

*Confidence interval.* $\P_\theta(\theta \in C_n) \ge 1-\alpha$ for all $\theta$.

*Pointwise asymptotic confidence interval.* $\liminf_{n\to\infty} \P_\theta(\theta \in C_n ) \ge 1- \alpha$ for all $\theta$ (e.g., the approximate normal-based confidence intervals).

*Uniform asymptotic confidence intercal.* $\liminf_{n\to\infty} \inf_{\theta\in\Theta}\P_\theta(\theta \in C_n ) \ge 1- \alpha$.

# 6.6 $\:$ Exercises
**Exercise 6.1.** Let $X_1,\dots, X_n \sim \Pois(\lambda)$ be IID. Let $\hat\lambda = \frac 1n\sum_{i=1}^n X_i$.

This estimator is unbiased. The standard error is $\sqrt{\frac \lambda n}$. The mean squared error is $\bias^2 +\se^2 = \frac \lambda n$.

**Exercise 6.2.** Let $X_1,\dots, X_n \sim \Uni([0,\theta])$ be IID. Let $\hat\theta = \max(X_1,\dots,X_n)$.

Noting that
$$ \E \hat\theta = \theta \left(1 - \frac 1{n+1} \right) = \frac{\theta n}{n+1},$$
we see that $\bias = - \frac{\theta}{n+1} $.

To compute $\var {\hat\theta}$, observe that the CDF of $\hat \theta$ is $(\frac t\theta)^n$ for $t \in [0,\theta]$. Then
$$
\begin{align*}
\E \hat\theta^2 &= \int_0^\theta t^2 \cdot n \frac{t^{n-1}}{\theta^n} \, dt \\
&= \frac{n}{n+2} \theta^2.
\end{align*}
$$

Then
$$
\begin{align*}
\var \hat\theta &= \E\hat\theta^2 - (\E \hat \theta)^2 \\
&= \theta^2 \left ( \frac{n}{n+2} - \frac{n^2}{(n+1)^2} \right ) \\
&= \frac{n\theta^2}{(n+2)(n+1)^2} \\
\se &= \sqrt\frac{n}{n+2}\frac{\theta}{n+1}.
\end{align*}
$$
The MSE is
$$ \bias^2 + \se^2 = \theta^2 \left ( \frac{1}{(n+1)^2} + \frac{n}{(n+2)(n+1)^2}\right) = \frac{\theta^2}{n+2}. $$
